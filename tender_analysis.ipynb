```python
import pandas as pd

df = pd.read_excel("main_2026.xlsx")

print("სტრიქონები და სვეტები:", df.shape)
df.head()
```

```python
cpv = df["მთავარი Cpv"].astype(str)

dazgveva = df[cpv.str.contains("(665", regex=False)]

print("დაზღვევის ტენდერები:", len(dazgveva))
```

```python
aghtsera = df["დამატებითი ინფორმაცია"].astype(str)

# 1. ძებნა საკვანძო სიტყვებით
sitkvebi = "ქონების შეფასებ|აქტივ[ა-ჰ]* შეფასებ|საშემფასებლო|საბაზრო ღირებულ"
shefaseba = df[aghtsera.str.contains(sitkvebi)]

# 2. ზედმეტი CPV კოდების ამოღება (ტრენინგი, საინჟინრო, არქიტექტურული)
gamoricxva = "80500000|71300000|71200000"
shefaseba = shefaseba[~shefaseba["მთავარი Cpv"].astype(str).str.contains(gamoricxva)]

# 3. ფასის ადეკვატურობის ტენდერების ამოღება
shefaseba = shefaseba[~shefaseba["დამატებითი ინფორმაცია"].astype(str).str.contains("ფასწარმოქმნის ადეკვატურ")]

# 4. ჩაშლილების წილი (ამოღებამდე ვითვლით!)
chashlili = ["არ შედგა", "დასრულებულია უარყოფითი შედეგით", "შეწყვეტილია"]
statistika = {}
sul = len(shefaseba)
chashlili_raod = shefaseba["სტატუსი"].isin(chashlili).sum()
print("შეფასება - სულ:", sul, "| ჩაშლილი:", chashlili_raod, "| წილი:", round(chashlili_raod / sul * 100), "%")
statistika["ქონების შეფასება"] = {"სულ": sul, "ჩაშლილი": chashlili_raod}

# 5. ჩაშლილების ამოღება
shefaseba = shefaseba[~shefaseba["სტატუსი"].isin(chashlili)]
print("შეფასება (საბოლოო):", len(shefaseba))
```

```python
# დაზღვევა: ჩაშლილების წილი (ამოღებამდე ვითვლით!)
sul = len(dazgveva)
chashlili_raod = dazgveva["სტატუსი"].isin(chashlili).sum()
print("დაზღვევა - სულ:", sul, "| ჩაშლილი:", chashlili_raod, "| წილი:", round(chashlili_raod / sul * 100), "%")
statistika["დაზღვევა"] = {"სულ": sul, "ჩაშლილი": chashlili_raod}

# დაზღვევა: ჩაშლილების ამოღება
dazgveva = dazgveva[~dazgveva["სტატუსი"].isin(chashlili)]
print("დაზღვევა (საბოლოო):", len(dazgveva))
```

```python
# 1. ყველა ტენდერი CPV 79200000-ით
audit = df[cpv.str.contains("(79200000", regex=False)]

# 2. შეფასების ტენდერების ამოღება (ისინი უკვე პირველ კატეგორიაშია)
audit = audit[~audit["დამატებითი ინფორმაცია"].astype(str).str.contains(sitkvebi)]

# 3. ფასის ადეკვატურობის ტენდერების ამოღება
audit = audit[~audit["დამატებითი ინფორმაცია"].astype(str).str.contains("ფასწარმოქმნის ადეკვატურ")]

# 4. ჩაშლილების წილი (ამოღებამდე)
sul = len(audit)
chashlili_raod = audit["სტატუსი"].isin(chashlili).sum()
print("აუდიტი - სულ:", sul, "| ჩაშლილი:", chashlili_raod, "| წილი:", round(chashlili_raod / sul * 100), "%")
statistika["აუდიტი და ბუღალტერია"] = {"სულ": sul, "ჩაშლილი": chashlili_raod}

# 5. ჩაშლილების ამოღება
audit = audit[~audit["სტატუსი"].isin(chashlili)]
print("აუდიტი (საბოლოო):", len(audit))

# 6. ISO და სერტიფიკაციის აუდიტების ამოღება (ფინანსური აუდიტი არ არის)
audit = audit[~audit["დამატებითი ინფორმაცია"].astype(str).str.contains("ISO|SWIFT|წარმოშობის სერთიფიკ")]
print("აუდიტი (ISO-ს გარეშე):", len(audit))
```

```python
# 1. თითოეულ ცხრილს კატეგორიის სვეტს ვუმატებთ
shefaseba = shefaseba.copy()
dazgveva = dazgveva.copy()
audit = audit.copy()

shefaseba["კატეგორია"] = "ქონების შეფასება"
dazgveva["კატეგორია"] = "დაზღვევა"
audit["კატეგორია"] = "აუდიტი და ბუღალტერია"

# 2. სამივეს ერთ ცხრილად ვაერთიანებთ
tenderebi = pd.concat([shefaseba, dazgveva, audit])
# შესწორებები: ადამიანის გადაწყვეტილება AI-სთან შედარების შემდეგ
shesworebebi = {
    659869: "ქონების შეფასება",
    655158: "ქონების შეფასება",
    654734: "ქონების შეფასება",
    653707: "ქონების შეფასება",
    672762: "ქონების შეფასება",
    677900: "აუდიტი და ბუღალტერია",
}
for tid, kat in shesworebebi.items():
    if tid in tenderebi["ID"].values:
        tenderebi.loc[tenderebi["ID"] == tid, "კატეგორია"] = kat
    else:
        axali = df[df["ID"] == tid].copy()
        axali["კატეგორია"] = kat
        tenderebi = pd.concat([tenderebi, axali])

# ჩაშლილების სტატისტიკის განახლება ახალი რაოდენობებით
for kat in statistika:
    statistika[kat]["სულ"] = (tenderebi["კატეგორია"] == kat).sum() + statistika[kat]["ჩაშლილი"]
print("სულ ტენდერი:", len(tenderebi))

# 3. შემოწმება: ხომ არ მოხვდა რომელიმე ტენდერი ორჯერ
print("გამეორებული ID:", tenderebi["ID"].duplicated().sum())

# 4. მიმოხილვა: კატეგორია × სტატუსი
pd.crosstab(tenderebi["სტატუსი"], tenderebi["კატეგორია"], margins=True, margins_name="სულ")
```

```python
# მხოლოდ დადებული ხელშეკრულებები, მხოლოდ ლარში
kontraktebi = tenderebi[(tenderebi["სტატუსი"] == "ხელშეკრულება დადებულია") & (tenderebi["ვალუტა"] == "GEL")].copy()
print("ხელშეკრულებები (ლარში):", len(kontraktebi))

# 1. ბაზრის მოცულობა: რამდენი ფული ტრიალებს
moculoba = kontraktebi.groupby("კატეგორია").agg(
    ხელშეკრულება=("ID", "count"),
    ჯამი_ლარი=("საბოლოო ღირებულება", "sum")
).round(0)
print(moculoba)

# 2. ფასის კლება: მხოლოდ იქ, სადაც ორივე ციფრი არსებობს
orive = kontraktebi.dropna(subset=["სავარაუდო ღირებულება"])
kleba = orive.groupby("კატეგორია")[["სავარაუდო ღირებულება", "საბოლოო ღირებულება"]].sum()
kleba["კლება_%"] = ((1 - kleba["საბოლოო ღირებულება"] / kleba["სავარაუდო ღირებულება"]) * 100).round(1)
kleba
```

```python
# 1. კომპანიების სახელების გასუფთავება (ზედმეტი ჰარების მოშორება)
kontraktebi["მიმწოდებელი"] = kontraktebi["მიმწოდებლის სახელი"].astype(str).str.split().str.join(" ")

# 2. თითოეულ კატეგორიაში ტოპ 5 კომპანია თანხის მიხედვით
for kategoria, jgufi in kontraktebi.groupby("კატეგორია"):
    gamarjvebulebi = jgufi.groupby("მიმწოდებელი").agg(
        მოგება=("ID", "count"),
        ჯამი_ლარი=("საბოლოო ღირებულება", "sum")
    ).sort_values("ჯამი_ლარი", ascending=False)
    gamarjvebulebi["წილი_%"] = (gamarjvebulebi["ჯამი_ლარი"] / gamarjvebulebi["ჯამი_ლარი"].sum() * 100).round(1)

    print("=====", kategoria, "| სულ კომპანია:", len(gamarjvebulebi))
    print(gamarjvebulebi.head(5).round(0))
    print()
```

```python
# 1. ერთმონაწილიანი ტენდერების წილი კატეგორიების მიხედვით
kontraktebi["ერთი_მონაწილე"] = kontraktebi["ბიდერების რაოდენობა"] <= 1

konkurencia = kontraktebi.groupby("კატეგორია").agg(
    ხელშეკრულება=("ID", "count"),
    ერთი_მონაწილით=("ერთი_მონაწილე", "sum"),
    საშუალო_მონაწილე=("ბიდერების რაოდენობა", "mean")
)
konkurencia["ერთის_წილი_%"] = (konkurencia["ერთი_მონაწილით"] / konkurencia["ხელშეკრულება"] * 100).round(0)
print(konkurencia.round(1))
print()

# 2. ფასის კლება: 1 მონაწილე vs 2+ მონაწილე
orive = kontraktebi.dropna(subset=["სავარაუდო ღირებულება"]).copy()
orive["ჯგუფი"] = orive["ერთი_მონაწილე"].map({True: "1 მონაწილე", False: "2+ მონაწილე"})

shedareba = orive.groupby("ჯგუფი").agg(
    ტენდერი=("ID", "count"),
    სავარაუდო=("სავარაუდო ღირებულება", "sum"),
    საბოლოო=("საბოლოო ღირებულება", "sum")
)
shedareba["კლება_%"] = ((1 - shedareba["საბოლოო"] / shedareba["სავარაუდო"]) * 100).round(1)
shedareba
```

```python
# 1. შემსყიდველების სახელების გასუფთავება
kontraktebi["შემსყიდველი_სუფთა"] = kontraktebi["შემსყიდველი"].astype(str).str.split().str.join(" ")

# 2. ტოპ 10 შემსყიდველი თანხის მიხედვით
shemskidvelebi = kontraktebi.groupby("შემსყიდველი_სუფთა").agg(
    ხელშეკრულება=("ID", "count"),
    ჯამი_ლარი=("საბოლოო ღირებულება", "sum")
).sort_values("ჯამი_ლარი", ascending=False)

shemskidvelebi["წილი_%"] = (shemskidvelebi["ჯამი_ლარი"] / shemskidvelebi["ჯამი_ლარი"].sum() * 100).round(1)
print("სულ შემსყიდველი:", len(shemskidvelebi))
print(shemskidvelebi.head(10).round(0))
print()

# 3. ტენდერები თვეების მიხედვით (გამოცხადების თარიღით)
kontraktebi["თვე"] = pd.to_datetime(kontraktebi["გამოცხადების დრო"]).dt.month
pd.crosstab(kontraktebi["თვე"], kontraktebi["კატეგორია"], margins=True, margins_name="სულ")
```

```python
# 1. ერთი მონაწილე (ეს სვეტი უკვე გვაქვს ნაბიჯი 9-დან)
kontraktebi["🚩_1_მონაწილე"] = kontraktebi["ერთი_მონაწილე"]

# 2. განმეორებადი წყვილი: შემსყიდველი + მიმწოდებელი 2-ჯერ ან მეტჯერ
tsyvili = kontraktebi.groupby(["შემსყიდველი_სუფთა", "მიმწოდებელი"])["ID"].transform("count")
kontraktebi["🚩_განმეორებადი"] = tsyvili >= 2

# 3. მცირე კლება: 5%-ზე ნაკლები (მხოლოდ სადაც სავარაუდო ღირებულება არსებობს)
kleba_wili = 1 - kontraktebi["საბოლოო ღირებულება"] / kontraktebi["სავარაუდო ღირებულება"]
kontraktebi["🚩_მცირე_კლება"] = kleba_wili < 0.05

# 4. რამდენი ნიშანი აქვს თითოეულ ხელშეკრულებას
nishnebi = ["🚩_1_მონაწილე", "🚩_განმეორებადი", "🚩_მცირე_კლება"]
kontraktebi["ნიშნების_რაოდენობა"] = kontraktebi[nishnebi].sum(axis=1)

print("თითოეული ნიშანი:")
print(kontraktebi[nishnebi].sum())
print()
pd.crosstab(kontraktebi["კატეგორია"], kontraktebi["ნიშნების_რაოდენობა"], margins=True, margins_name="სულ")
```

```python
import json, re, time

# 1. ფართო წრე: ყველა, ვინც შეიძლება ჩვენს სამ კატეგორიას ეკუთვნოდეს
cpv_all = df["მთავარი Cpv"].astype(str)
aghtsera_all = df["დამატებითი ინფორმაცია"].astype(str)
pirobi = cpv_all.str.contains(r"\((?:665|792|793|794)") | aghtsera_all.str.contains("შეფასებ|აუდიტ|დაზღვევ|ბუღალტ|საშემფასებლო|საბაზრო ღირებულ")
wre = df[pirobi & ~df["სტატუსი"].isin(chashlili)].copy()
print("შესამოწმებელი ტენდერები:", len(wre))

# 2. ინსტრუქცია AI-სთვის (ჩვენი წესები)
INSTRUQCIA = """შენ ხარ სახელმწიფო შესყიდვების ექსპერტი. თითოეულ ტენდერს მიანიჭე ზუსტად ერთი კატეგორია:
- შეფასება: უძრავი/მოძრავი ქონების, აქტივების, მიწის ან ბიზნესის ღირებულების შეფასება (მაშინაც, თუ „აუდიტორული მომსახურება“ ჰქვია).
- დაზღვევა: დაზღვევის მომსახურების შესყიდვა (ჯანმრთელობის, ქონების, ავტომობილის, პასუხისმგებლობის და ა.შ.).
- აუდიტი: ფინანსური აუდიტი, ბუღალტრული მომსახურება, ინვენტარიზაცია.
- სხვა: ყველაფერი დანარჩენი, მათ შორის ISO/ხარისხის სერტიფიკაციის აუდიტი, ფასის ადეკვატურობის ექსპერტიზა, შემფასებლების სწავლება, მშენებლობა, საქონელი.

უპასუხე მხოლოდ JSON ობიექტით, სხვა ტექსტის გარეშე, ასეთი ფორმით: {"ID": "კატეგორია", "ID": "კატეგორია"}"""

def klasifikacia(paketi):
    teqsti = "\n".join(
        f'{r["ID"]}: ' + " ".join(str(r["დამატებითი ინფორმაცია"]).split())[:400]
        for _, r in paketi.iterrows()
    )
    pasuxi = ai.generate_text(INSTRUQCIA + "\n\nტენდერები:\n" + teqsti, model_name="google/gemini-3.6-flash")
    m = re.search(r"\{.*\}", pasuxi, re.S)
    return json.loads(m.group(0)) if m else {}

# 3. 20-20-ად გაგზავნა
ai_shedegi = {}
zoma = 20
for i in range(0, len(wre), zoma):
    paketi = wre.iloc[i:i + zoma]
    try:
        ai_shedegi.update(klasifikacia(paketi))
        print(f"პაკეტი {i // zoma + 1}: მზადაა ({len(ai_shedegi)} ტენდერი)")
    except Exception as e:
        print(f"პაკეტი {i // zoma + 1}: შეცდომა – {e}")
    time.sleep(2)

wre["AI_კატეგორია"] = wre["ID"].astype(str).map(ai_shedegi)
print("\nAI-ის პასუხი არ აქვს:", wre["AI_კატეგორია"].isna().sum())
wre["AI_კატეგორია"].value_counts()
```
